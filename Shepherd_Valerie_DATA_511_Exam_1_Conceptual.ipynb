# Exercise 6 (5 points)

Explain the concept of **feasible region** in linear programming and its significance in the solution process. Be specific.

The feasible region is the set of all possible combinations of decision variables that satisfy all of the constraints in a linear programming problem, including non-negativity constraints. In a linear programming (LP) model, the feasible region is the set of all combinations of decision variables that satisfy all constraints. In an integer programming (IP) model, the feasible region is also restricted so that the required decision variables must take integer values. # Exercise 11 (6 points)

A factory has two machines, Machine A and Machine B. Machine A produces 40% of the total items, and Machine B produces 60%. The defect rate is 2% for Machine A and 5% for Machine B. If a randomly selected item is defective, what is the probability it was produced by Machine B?

Using Bayes' Theorem:

P(B|D) = [P(D|B) * P(B)] / [P(D|A) * P(A) + P(D|B) * P(B)]

P(B|D) = [(0.05)(0.60)] / [(0.02)(0.40) + (0.05)(0.60)]

P(B|D) = 0.03 / (0.008 + 0.03)

P(B|D) = 0.03 / 0.038

P(B|D) = 0.7895

Therefore, the probability that a defective item was produced by Machine B is approximately 78.95%.

# Exercise 15 (5 points)

Define the concept of **expected value (EV)** in the context of decision analysis. Explain how it is used to compare decision alternatives under uncertainty. Be specific.

Expected value (EV) is the probability-weighted average of the possible outcomes of a decision. It is calculated by multiplying each possible outcome by its probability and adding the results. In decision analysis, EV is used to compare alternatives under uncertainty. The decision with the highest expected value is generally preferred when the goal is to maximize expected payoff.

# Exercise 16 (5 points)

What is **sensitivity analysis** in the context of linear programming, and why is it important for decision makers? Be specific.

Sensitivity analysis in linear programming examines how changes in the model's inputs, such as objective function coefficients, resource availability, or constraints, affect the optimal solution. It is important because it helps decision makers understand how stable the optimal solution is and how changes in business conditions could affect the decision. For example, sensitivity analysis can show how much a resource constraint or product profit can change before the current optimal solution changes.

# Exercise 20 (5 points)

Describe the main steps of a **Monte Carlo simulation** and explain why it is useful for decision-making under uncertainty. Be specific.

The main steps of a Monte Carlo simulation are to identify the uncertain variables, assign probability distributions to those variables, generate random values based on the distributions, and run the model many times. # Exercise 23 (5 points)

Identify and define the three main components of a linear programming model: **decision variables**, **objective function**, and **constraints**. Illustrate each with a simple production planning example.

The three main components of a linear programming model are decision variables, the objective function, and constraints.

Decision variables are the quantities that the decision maker must determine. For example, a company could define x as the number of Product A units produced and y as the number of Product B units produced.

The objective function specifies what the company wants to maximize or minimize. For example, if Product A earns $30 profit and Product B earns $20 profit, the objective could be:

Maximize Profit = 30x + 20y

Constraints are limitations on the decision variables, such as available labor, materials, or production capacity. For example:

2x + y <= 100

This could represent a limit of 100 hours of available labor.

Together, the decision variables, objective function, and constraints form the linear programming model used to determine the optimal production plan.
